```python
room = ["D", "C", "D", "D", "C"]
def show_room(room):
  picture = ""
  for spot in room:
    if spot == "D":
      picture += "💩"
  else:
    picture += "✨"
  print(picture)
print("our room right now")
show_room(room)
```

```python
def clean_spot(spot):
    if spot == "D":
      return "C"
    else:
      return "C"

result = clean_spot("D")
print("The robot looked at a dirty sport 💩 and made it:", result, "(c means clean ✨)")
```

```python

print("Before - The dirty room:",)
show_room(room)
print()

for i in range(len(room)):
  room[i] = clean_spot(room[i])
  print("After cleaning spot number", i+1, ":")
  show_room(room)
  print()
  print("After - all done! ✨🤖✨")
  show_room(room)

```
